# Handwritten Digit Recognition on MNIST (Keras)

## 1. Load the MNIST dataset and display sample images

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

mnist = tf.keras.datasets.mnist
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

print("Number of training images:", len(train_images))
print("Number of test images    :", len(test_images))
print("Shape of one image       :", train_images[0].shape)
print("Digits (classes)         :", np.unique(train_labels))

fig, ax = plt.subplots(2, 5, figsize=(10, 4.5))
for i, a in enumerate(ax.flat):
    a.imshow(train_images[i], cmap="gray")
    a.set_title(f"Label: {train_labels[i]}")
    a.axis("off")
plt.show()

# normalise pixel values to the range 0-1
train_images = train_images / 255.0
test_images = test_images / 255.0
```

## 2. Design, compile and train the neural network

```python
model = tf.keras.Sequential()
model.add(tf.keras.layers.Input(shape=(28, 28)))
model.add(tf.keras.layers.Flatten())
model.add(tf.keras.layers.Dense(128, activation="relu"))
model.add(tf.keras.layers.Dense(10, activation="softmax"))

model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])

model.summary()

history = model.fit(train_images, train_labels,
                    epochs=10,
                    validation_split=0.2)
```

## 3. Evaluate the model using test accuracy

```python
loss, accuracy = model.evaluate(test_images, test_labels)
print(f"Test accuracy: {accuracy:.4f}")
print(f"Test loss    : {loss:.4f}")
```

## 4. Visualize training/validation accuracy and loss

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.plot(history.history["accuracy"], label="Training")
ax1.plot(history.history["val_accuracy"], label="Validation")
ax1.set_title("Accuracy")
ax1.set_xlabel("Epoch")
ax1.legend()

ax2.plot(history.history["loss"], label="Training")
ax2.plot(history.history["val_loss"], label="Validation")
ax2.set_title("Loss")
ax2.set_xlabel("Epoch")
ax2.legend()

plt.show()
```

## 5. Test the model on 5 handwritten images (actual vs predicted)

```python
sample_images = test_images[:5]
sample_actual = test_labels[:5]
sample_predicted = np.argmax(model.predict(sample_images), axis=1)

fig, ax = plt.subplots(1, 5, figsize=(10, 3))
for i in range(5):
    ax[i].imshow(sample_images[i], cmap="gray")
    ax[i].set_title(f"Actual: {sample_actual[i]}\nPredicted: {sample_predicted[i]}")
    ax[i].axis("off")
plt.show()
```

## 6. Experiment: 256 neurons + Dropout (0.3)

```python
model_new = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28, 28)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(256, activation="relu"),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(10, activation="softmax")
])

model_new.compile(optimizer="adam",
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])

history_new = model_new.fit(train_images, train_labels,
                            epochs=10,
                            validation_split=0.2)

loss_new, accuracy_new = model_new.evaluate(test_images, test_labels)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.plot(history_new.history["accuracy"], label="Training")
ax1.plot(history_new.history["val_accuracy"], label="Validation")
ax1.set_title("Accuracy (256 neurons + Dropout)")
ax1.set_xlabel("Epoch")
ax1.legend()

ax2.plot(history_new.history["loss"], label="Training")
ax2.plot(history_new.history["val_loss"], label="Validation")
ax2.set_title("Loss (256 neurons + Dropout)")
ax2.set_xlabel("Epoch")
ax2.legend()

plt.show()

print(f"Baseline model  -> Test accuracy: {accuracy:.4f}  Test loss: {loss:.4f}")
print(f"Modified model  -> Test accuracy: {accuracy_new:.4f}  Test loss: {loss_new:.4f}")
```
